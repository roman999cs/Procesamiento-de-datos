<h2 id="tuplas">Tuplas</h2>

Una tupla es una secuencia ordenada e <b>inmutable</b> de elementos. Se define entre paréntesis y los elementos van separados por comas.

```python
Partido = ('Millonarios', 'Nacional', 2, 1)
Partido
```

El tipo de dato de una tupla es <code>tuple</code>:

```python
type(Partido)
```

Una tupla puede contener distintos tipos de datos: cadenas, enteros, flotantes, etc.

```python
Jugador = ('Falcao', 9, 1.79, 'Delantero')
Jugador
```

<h2 id="index">Indexación</h2>

Cada elemento de una tupla corresponde a un índice, empezando en 0.

```python
Partido[0]
```

```python
Partido[2]
```

También podemos usar índices negativos para contar desde el final:

```python
Partido[-1]
```

<h2 id="concat">Concatenación de Tuplas</h2>

Podemos concatenar varias tuplas con el operador <code>+</code>, por ejemplo para armar el resumen de la fecha:

```python
Fecha1 = ('Millonarios', 'Nacional', 2, 1)
Fecha2 = ('Junior', 'América', 0, 0)
Resumen = Fecha1 + Fecha2
Resumen
```

```python
len(Resumen)
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Aquí concatené dos tuplas de resultados con + para armar un resumen de fecha más grande, y conté cuántos elementos quedaron con len().</p>
</div>

<h2 id="slice">Slicing (Sublistas)</h2>

Podemos obtener varios elementos de una tupla usando slicing, indicando el índice inicial y el final (sin incluir este último).

```python
Resumen[0:2]
```

```python
Resumen[2:]
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> En esta parte usé slicing para sacar sub-tuplas del resumen de resultados, sin modificar la tupla original.</p>
</div>

<h2 id="sort">Ordenamiento</h2>

Supongamos que tenemos los goles anotados por un jugador en las últimas 5 fechas, en una tupla.

```python
Goles = (2, 0, 1, 3, 0)
Goles
```

Podemos usar la función <code>sorted()</code> para ordenar los elementos de menor a mayor. Esto devuelve una <b>lista</b> nueva, ya que la tupla original no se puede modificar.

```python
GolesOrdenados = sorted(Goles)
GolesOrdenados
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Acá entendí que sorted() no modifica la tupla original sino que devuelve una lista nueva ordenada, porque las tuplas no se pueden cambiar.</p>
</div>

<h2 id="nested">Tupla Anidada</h2>

Una tupla puede contener otras tuplas dentro, por ejemplo la tabla de posiciones con equipo, puntos y una tupla de (ganados, empatados, perdidos):

```python
Posicion1 = ('Nacional', 30, (9, 3, 2))
Posicion1
```

Podemos acceder a los elementos anidados encadenando índices:

```python
Posicion1[2]
```

```python
Posicion1[2][0]
```

```python
Posicion1[2][0][0] if isinstance(Posicion1[2][0], tuple) else Posicion1[2][0]
```
