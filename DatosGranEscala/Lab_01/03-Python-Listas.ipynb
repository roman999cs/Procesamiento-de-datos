<h2 id="listas">Listas</h2>

Una lista es una secuencia ordenada y <b>mutable</b> de elementos, definida entre corchetes <code>[]</code>.

```python
Plantilla = ['Ospina', 'Cuadrado', 'Muñoz', 'Lerma', 'Díaz']
Plantilla
```

Una lista puede mezclar distintos tipos de datos:

```python
Jugador = ['James Rodríguez', 10, 1.80, 'Volante']
Jugador
```

Las listas pueden estar anidadas, por ejemplo una lista de listas con varios jugadores:

```python
Nomina = [['Ospina', 1, 'Portero'], ['Cuadrado', 15, 'Defensa'], ['Díaz', 7, 'Delantero']]
Nomina
```

<h2 id="index">Indexación</h2>

Al igual que con las cadenas y tuplas, cada elemento tiene un índice, empezando en 0.

```python
Plantilla[0]
```

```python
Plantilla[-1]
```

Podemos usar slicing para obtener varios elementos:

```python
Plantilla[1:3]
```

<h2 id="content">Contenido de la Lista</h2>

A diferencia de las tuplas, las listas son <b>mutables</b>: podemos cambiar sus elementos.

```python
Plantilla[0] = 'Vargas'
Plantilla
```

Podemos concatenar listas con <code>+</code>, por ejemplo para sumar refuerzos a la plantilla:

```python
Refuerzos = ['Borré', 'Zapata']
Plantilla + Refuerzos
```

<h2 id="ops">Operaciones con Listas</h2>

El método <code>append()</code> agrega un elemento al final de la lista:

```python
Goleadores = ['Falcao', 'Muriel']
Goleadores.append('Zapata')
Goleadores
```

El método <code>extend()</code> agrega varios elementos de otra lista:

```python
Goleadores.extend(['Borré', 'Díaz'])
Goleadores
```

El método <code>del</code> elimina un elemento dado su índice:

```python
del(Goleadores[0])
Goleadores
```

Podemos convertir una cadena separada por comas en una lista con <code>split()</code>, por ejemplo la alineación titular:

```python
Titulares = 'Ospina,Cuadrado,Muñoz,Lerma,Díaz'
Titulares.split(',')
```

<h2 id="copy">Copiado y Clonación de Listas</h2>

Si asignamos una lista a una nueva variable con <code>=</code>, ambas variables apuntan a la <b>misma</b> lista en memoria.

```python
EquipoA = ['Ospina', 'Cuadrado', 'Muñoz']
EquipoB = EquipoA
EquipoB[0] = 'Vargas'
EquipoA
```

Para clonar una lista de verdad (una copia independiente), usamos slicing completo <code>[:]</code>:

```python
EquipoC = EquipoA[:]
EquipoC[0] = 'Otro Portero'
print('EquipoA:', EquipoA)
print('EquipoC:', EquipoC)
```
