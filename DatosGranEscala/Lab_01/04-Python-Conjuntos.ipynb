<h2 id="sets">Conjuntos</h2>

Un conjunto es una colección <b>no ordenada</b> de elementos <b>únicos</b>, definida entre llaves <code>{}</code>.

```python
Clasificados = {'Argentina', 'Brasil', 'Colombia', 'Uruguay', 'Colombia'}
Clasificados
```

Nota que el elemento repetido <code>'Colombia'</code> solo aparece una vez: los conjuntos eliminan duplicados automáticamente.

También podemos crear un conjunto a partir de una lista usando <code>set()</code>:

```python
Lista = ['Real Madrid', 'Barcelona', 'Real Madrid', 'Sevilla']
EquiposUnicos = set(Lista)
EquiposUnicos
```

<h2 id="content">Contenido de un Conjunto</h2>

Podemos agregar un elemento con <code>add()</code>:

```python
Clasificados.add('Ecuador')
Clasificados
```

Si intentamos agregar un elemento que ya existe, el conjunto no cambia:

```python
Clasificados.add('Colombia')
Clasificados
```

Podemos eliminar un elemento con <code>remove()</code>:

```python
Clasificados.remove('Uruguay')
Clasificados
```

Con <code>in</code> verificamos si un elemento pertenece al conjunto:

```python
'Brasil' in Clasificados
```

<h2 id="ops">Operaciones con Conjuntos</h2>

Supongamos dos conjuntos: equipos que jugaron Copa América y equipos que jugaron Copa Libertadores.

```python
CopaAmerica = {'Argentina', 'Colombia', 'Brasil', 'Uruguay'}
Libertadores = {'Colombia', 'Brasil', 'Ecuador', 'Paraguay'}
```

La <b>intersección</b> (<code>&</code>) da los elementos que están en ambos conjuntos:

```python
CopaAmerica & Libertadores
```

La <b>unión</b> (<code>|</code>) da todos los elementos de ambos conjuntos, sin repetir:

```python
CopaAmerica | Libertadores
```

La <b>diferencia</b> (<code>-</code>) da los elementos que están en un conjunto pero no en el otro:

```python
CopaAmerica - Libertadores
```

<h2 id="logic">Operacion Lógicas con Conjuntos</h2>

Podemos verificar si un conjunto es <b>subconjunto</b> de otro con <code>issubset()</code>:

```python
Semifinalistas = {'Argentina', 'Colombia'}
Semifinalistas.issubset(CopaAmerica)
```

También podemos verificar si dos conjuntos son <b>disjuntos</b> (no comparten elementos) con <code>isdisjoint()</code>:

```python
GrupoA = {'Argentina', 'Chile'}
GrupoB = {'Colombia', 'Brasil'}
GrupoA.isdisjoint(GrupoB)
```
