<h2 id="dicc">Diccionarios</h2>

Un diccionario almacena datos en pares <b>llave: valor</b>, entre llaves <code>{}</code>.

```python
Jugador = {'Nombre': 'James Rodríguez', 'Dorsal': 10, 'Posición': 'Volante', 'Goles': 5}
Jugador
```

<h2 id="que-son">¿Qué son los diccionarios?</h2>

A diferencia de las listas, en un diccionario accedemos a los valores usando su <b>llave</b>, no un índice numérico.

```python
Jugador['Nombre']
```

```python
Jugador['Goles']
```

Un diccionario también puede tener listas o incluso otros diccionarios como valores, por ejemplo el historial de resultados de un equipo:

```python
Equipo = {'Nombre': 'Nacional', 'Resultados': ['G', 'E', 'G', 'P'], 'DT': 'Un técnico'}
Equipo['Resultados']
```

<h2 id="llaves">Llaves</h2>

Podemos obtener todas las llaves de un diccionario con <code>keys()</code>:

```python
Jugador.keys()
```

Y todos los valores con <code>values()</code>:

```python
Jugador.values()
```

Para agregar un nuevo par llave-valor, simplemente asignamos un valor a una llave nueva:

```python
Jugador['Asistencias'] = 3
Jugador
```

Para eliminar una llave, usamos <code>del</code>:

```python
del(Jugador['Asistencias'])
Jugador
```

Con <code>in</code> podemos verificar si una llave existe en el diccionario:

```python
'Dorsal' in Jugador
```
