<h2 id="declaraciones">Declaraciones de Condición</h2>

Las condiciones nos permiten ejecutar código solo si se cumple algo, por ejemplo, si un equipo ganó un partido.

```python
GolesLocal = 2
GolesVisitante = 1
GolesLocal > GolesVisitante
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Vi que una condición en Python es simplemente una comparación que devuelve True o False, como comparar los goles de dos equipos.</p>
</div>

<h2 id="comparacion">Operadores de Comparación</h2>

Python tiene varios operadores de comparación: <code>&gt;</code>, <code>&lt;</code>, <code>&gt;=</code>, <code>&lt;=</code>, <code>==</code> y <code>!=</code>.

```python
GolesLocal == GolesVisitante
```

```python
GolesLocal != GolesVisitante
```

También podemos comparar cadenas, por ejemplo dos nombres de equipos:

```python
'Nacional' == 'Millonarios'
```

<h2 id="branching">Sentencias Condicionales (Branching)</h2>

Con <code>if</code>, <code>elif</code> y <code>else</code> podemos ramificar el código según el resultado de un partido.

```python
GolesLocal = 2
GolesVisitante = 1

if GolesLocal > GolesVisitante:
    print("Gana el equipo local")
elif GolesLocal < GolesVisitante:
    print("Gana el equipo visitante")
else:
    print("Empate")
```

Podemos anidar condiciones, por ejemplo para decidir si un jugador es expulsado según sus tarjetas amarillas:

```python
TarjetasAmarillas = 2

if TarjetasAmarillas >= 2:
    print("El jugador es expulsado")
else:
    print("El jugador sigue en cancha")
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Con if/elif/else armé la lógica típica de un partido: quién gana, quién pierde o si hay empate, y también un ejemplo anidado con tarjetas amarillas.</p>
</div>

<h2 id="logicos">Operadores Lógicos</h2>

El operador <code>and</code> devuelve <code>True</code> solo si ambas condiciones son verdaderas. Por ejemplo, para saber si un equipo clasifica necesitando ganar Y que otro resultado le favorezca:

```python
Gano = True
OtroResultadoFavorece = False
Gano and OtroResultadoFavorece
```

El operador <code>or</code> devuelve <code>True</code> si al menos una condición es verdadera:

```python
Gano or OtroResultadoFavorece
```

El operador <code>not</code> invierte el valor de una condición:

```python
not Gano
```
