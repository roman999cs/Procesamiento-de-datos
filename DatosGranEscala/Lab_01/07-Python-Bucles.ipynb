<h2 id="bucles">Bucles</h2>

Los bucles nos permiten repetir una acción varias veces, por ejemplo recorrer todos los jugadores de una plantilla.

<h2 id="rango">Rango</h2>

```python
list(range(5))
```

```python
list(range(1, 6))
```

<h2 id="for">¿Qué es el bucle <code>for</code>?</h2>

El bucle <code>for</code> recorre cada elemento de una secuencia. Recorramos la plantilla de un equipo:

```python
Plantilla = ['Ospina', 'Cuadrado', 'Muñoz', 'Lerma', 'Díaz']

for Jugador in Plantilla:
    print(Jugador)
```

También podemos usar <code>enumerate()</code> para obtener el índice junto al elemento, por ejemplo el dorsal según la posición en la lista:

```python
for i, Jugador in enumerate(Plantilla):
    print('Dorsal', i + 1, ':', Jugador)
```

Podemos sumar los goles de un jugador partido a partido usando un bucle <code>for</code>:

```python
GolesPorFecha = [1, 0, 2, 0, 1]
TotalGoles = 0

for g in GolesPorFecha:
    TotalGoles = TotalGoles + g

print('Goles totales:', TotalGoles)
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Practiqué el bucle for recorriendo la plantilla de un equipo, usé enumerate() para sacar el índice junto al jugador, y sumé los goles de un jugador partido a partido.</p>
</div>

<h2 id="while">¿Qué es el bucle <code>while</code>?</h2>

El bucle <code>while</code> se repite mientras una condición sea verdadera. Simulemos el conteo regresivo antes de un partido:

```python
Minuto = 5

while Minuto > 0:
    print('Faltan', Minuto, 'minutos para el pitazo inicial')
    Minuto = Minuto - 1

print('¡Comienza el partido!')
```

Otro ejemplo: recorrer los goles de un jugador hasta encontrar el primer partido en el que no anotó.

```python
GolesPorFecha = [1, 2, 0, 3, 1]
i = 0

while i < len(GolesPorFecha) and GolesPorFecha[i] > 0:
    print('Fecha', i + 1, ': anotó', GolesPorFecha[i], 'gol(es)')
    i = i + 1

print('Se cortó la racha en la fecha', i + 1)
```

<div style="background:#f6f8fa; border-left:4px solid #2c3e50; padding:8px 12px;">
<p>📝 <b>Mi documentación:</b> Con while simulé una cuenta regresiva antes de un partido y también recorrí los goles de un jugador hasta que se cortó la racha de gol.</p>

Usa un bucle <code>while</code> para sumar los goles en <code>GolesPorFecha</code> hasta que la suma supere 5, e imprime en cuántas fechas se logró.

```python
GolesPorFecha = [2, 1, 3, 0, 2]

suma = 0
fechas = 0
while suma <= 5:
    suma += GolesPorFecha[fechas]
    fechas += 1

print(fechas)
```
