<h2 id="funciones">Funciones</h2>

<h3 id="que-es">¿Qué es una función?</h3>

Una función es un bloque de código reutilizable que recibe entradas (parámetros) y puede devolver un resultado. Se define con <code>def</code>.

```python
def saludo_equipo(nombre):
    print('¡Vamos ' + nombre + '!')

saludo_equipo('Colombia')
```

Una función también puede devolver un valor con <code>return</code>, por ejemplo el promedio de goles por partido:

```python
def promedio_goles(goles, partidos):
    return goles / partidos

promedio_goles(18, 9)
```

<h3 id="variables">Variables</h3>

Los parámetros de una función son variables locales que solo existen dentro de ella.

```python
def es_goleador(goles):
    limite = 10
    return goles >= limite

es_goleador(12)
```

<h3 id="faciles">Las funciones hacen las cosas más fáciles</h3>

En vez de repetir el mismo cálculo para cada jugador, escribimos la función una vez y la reutilizamos.

```python
def puntos_equipo(ganados, empatados):
    return ganados * 3 + empatados

print(puntos_equipo(9, 3))
print(puntos_equipo(5, 6))
print(puntos_equipo(2, 2))
```

<h3 id="predefinidas">Funciones predefinidas</h3>

Python trae funciones predefinidas como <code>sum()</code>, <code>max()</code> y <code>len()</code>, útiles para tablas de goles.

```python
GolesPorFecha = [1, 0, 2, 3, 1]
print(sum(GolesPorFecha))
print(max(GolesPorFecha))
print(len(GolesPorFecha))
```

<h3 id="if-ciclos">Utilizando <code>if</code>/<code>else</code> y ciclos en Funciones</h3>

Podemos combinar condiciones y bucles dentro de una función, por ejemplo para clasificar el desempeño de un equipo según sus puntos:

```python
def clasificar_equipo(puntos):
    if puntos >= 30:
        return 'Clasifica a copa internacional'
    elif puntos >= 20:
        return 'Zona media'
    else:
        return 'Zona de descenso'

for p in [35, 22, 10]:
    print(clasificar_equipo(p))
```

<h3 id="default">Establecer valores de argumento predeterminados en sus funciones personalizadas</h3>

Podemos darle un valor por defecto a un parámetro, por ejemplo la duración estándar de un partido:

```python
def minutos_jugados(titular=True):
    if titular:
        return 90
    else:
        return 20

print(minutos_jugados())
print(minutos_jugados(False))
```

<h3 id="globales">Variable globales</h3>

Una variable definida fuera de cualquier función es <b>global</b> y se puede leer desde dentro de una función.

```python
TorneoActual = 'Liga BetPlay'

def mostrar_torneo():
    print('Estamos jugando:', TorneoActual)

mostrar_torneo()
```

<h3 id="ambito">Ámbito de una variable</h3>

Una variable creada dentro de una función es <b>local</b>: solo existe dentro de esa función.

```python
def resumen_partido():
    marcador = '2-1'
    print('Marcador del partido:', marcador)

resumen_partido()
```

```python
# Esto daría error porque 'marcador' no existe fuera de la función
try:
    print(marcador)
except NameError as e:
    print('Error:', e)
```
