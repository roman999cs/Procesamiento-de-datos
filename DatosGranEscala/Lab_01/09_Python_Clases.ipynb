<h2 id="intro">Introducción a las Clases y Objetos</h2>

En programación orientada a objetos, una <b>clase</b> es un molde para crear <b>objetos</b>. Por ejemplo, todos los jugadores de fútbol comparten atributos (nombre, dorsal, posición) y comportamientos (anotar gol, hacer una asistencia).

<h2 id="crear-clase-1">Crear una Clase</h2>

Definimos una clase con la palabra clave <code>class</code>. El método <code>__init__</code> se ejecuta al crear cada objeto y define sus atributos iniciales.

```python
class Jugador:
    def __init__(self, nombre, dorsal, posicion):
        self.nombre = nombre
        self.dorsal = dorsal
        self.posicion = posicion
        self.goles = 0
```

<h2 id="instancias">Instancias de la Clase: Objetos y Atributos</h2>

Una <b>instancia</b> es un objeto concreto creado a partir de la clase. Creemos un jugador:

```python
j1 = Jugador('Luis Díaz', 7, 'Delantero')
print(j1.nombre, j1.dorsal, j1.posicion)
```

Podemos acceder y modificar los atributos del objeto usando el punto <code>.</code>:

```python
j1.goles = 5
j1.goles
```

<h2 id="metodos">Métodos</h2>

Un <b>método</b> es una función definida dentro de la clase que opera sobre el objeto. Agreguemos un método para anotar goles:

```python
class Jugador:
    def __init__(self, nombre, dorsal, posicion):
        self.nombre = nombre
        self.dorsal = dorsal
        self.posicion = posicion
        self.goles = 0

    def anotar_gol(self):
        self.goles = self.goles + 1
        print(self.nombre, 'anotó un gol! Lleva', self.goles)
```

```python
j1 = Jugador('Luis Díaz', 7, 'Delantero')
j1.anotar_gol()
j1.anotar_gol()
```

<h2 id="crear-clase-2">Crear una Clase</h2>

Ahora definamos también una clase <code>Equipo</code> que agrupe varios jugadores.

```python
class Equipo:
    def __init__(self, nombre):
        self.nombre = nombre
        self.jugadores = []

    def fichar(self, jugador):
        self.jugadores.append(jugador)
        print(jugador.nombre, 'ficha por', self.nombre)

    def total_goles(self):
        total = 0
        for j in self.jugadores:
            total = total + j.goles
        return total
```

<h2 id="instancia-jugador">Crear una instancia de la clase Jugador</h2>

```python
james = Jugador('James Rodríguez', 10, 'Volante')
falcao = Jugador('Falcao García', 9, 'Delantero')
james.anotar_gol()
falcao.anotar_gol()
falcao.anotar_gol()
```

<h2 id="equipo">La Clase Equipo</h2>

```python
seleccion = Equipo('Selección Colombia')
seleccion.fichar(james)
seleccion.fichar(falcao)
print('Goles totales del equipo:', seleccion.total_goles())
```

```python
for j in seleccion.jugadores:
    print(j.nombre, '-', j.posicion, '-', j.goles, 'goles')
```
