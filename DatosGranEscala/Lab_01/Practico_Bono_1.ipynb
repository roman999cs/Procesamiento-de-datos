#### MENOR GOLEADOR DE LOS DOS: Escriba una función que devuelva el menor de dos totales de goles *si* ambos son números pares, pero devuelva el mayor si uno o ambos son impares.

    menor_de_dos_pares(2,4) --> 2
    menor_de_dos_pares(2,5) --> 5

```python
def menor_de_dos_pares(a, b):
    if a % 2 == 0 and b % 2 == 0:
        return min(a, b)
    return max(a, b)
```

```python
# Revisión
menor_de_dos_pares(2, 4)
```

```python
# Revisión
menor_de_dos_pares(2, 5)
```

#### MISMA INICIAL: Escribe una función que tome el nombre de un equipo (dos palabras) y devuelva <code>True</code> si ambas palabras empiezan con la misma letra.

    mismo_inicio('Deportivo Dinamo') --> True
    mismo_inicio('Deportes Rojos') --> False

```python
def mismo_inicio(texto):
    palabras = texto.split()
    return palabras[0][0].lower() == palabras[1][0].lower()
```

```python
# Revisión
mismo_inicio('Deportivo Dinamo')
```

```python
# Revisión
mismo_inicio('Deportes Rojos')
```

#### MARCADOR DE 20: Dados dos enteros que representan el total de espectadores (en miles) en dos tribunas, devuelve <code>True</code> si la suma es 20 o si uno de los dos valores es 20. En caso contrario devuelve <code>False</code>.

    hace_veinte(20,10) --> True
    hace_veinte(12,8) --> True
    hace_veinte(2,3) --> False

```python
def hace_veinte(n1, n2):
    return n1 + n2 == 20 or n1 == 20 or n2 == 20
```

#### NOMBRE DE JUGADOR: Escribe una función que ponga en mayúscula la primera letra del nombre y la primera letra del apellido de un jugador, dados en una sola palabra separados por guion bajo.

    formatear_jugador('james_rodriguez') --> JamesRodriguez

```python
def formatear_jugador(nombre):
    # separo por el guion bajo y pongo mayúscula inicial en cada parte, luego las uno sin separador
    partes = nombre.split('_')
    return ''.join(p.capitalize() for p in partes)
```

```python
# Revisión
formatear_jugador('james_rodriguez')
```

#### GRITO DE HINCHADA AL REVÉS: Dada una frase, retorna la misma frase pero con las palabras en orden inverso.

    grito_al_reves('vamos mi equipo') --> 'equipo mi vamos'
    grito_al_reves('gana el partido') --> 'partido el gana'

Nota: revisa el método <code>.join()</code>:

    >>> '-'.join(['a', 'b', 'c'])
    'a-b-c'

```python
def grito_al_reves(texto):
    palabras = texto.split()
    palabras.reverse()
    return ' '.join(palabras)
```

```python
# Revisión
grito_al_reves('vamos mi equipo')
```

```python
# Revisión
grito_al_reves('gana el partido')
```

#### DOBLETE SEGUIDO: Dada una lista con los goles anotados por un jugador en cada partido, retorna <code>True</code> si la lista tiene un 2 seguido inmediatamente por otro 2 (dos partidos seguidos anotando doblete).

    doblete_seguido([1, 2, 2]) → True
    doblete_seguido([1, 2, 1, 2]) → False
    doblete_seguido([2, 1, 2]) → False

```python
def doblete_seguido(goles):
    for i in range(len(goles) - 1):
        if goles[i] == 2 and goles[i + 1] == 2:
            return True
    return False
```

```python
# Revisión
doblete_seguido([1, 2, 2])
```

```python
# Revisión
doblete_seguido([1, 2, 1, 2])
```

```python
# Revisión
doblete_seguido([2, 1, 2])
```

#### ECO DE ESTADIO: Dada una cadena, devuelve una cadena donde cada carácter se repite tres veces seguidas, simulando el eco del estadio.

    eco_estadio('Gol') --> 'GGGooolll'
    eco_estadio('Vamos') --> 'VVVaaammmooosss'

```python
def eco_estadio(texto):
    resultado = ''
    for letra in texto:
        resultado += letra * 3
    return resultado
```

```python
# Revisión
eco_estadio('Gol')
```

```python
# Revisión
eco_estadio('Vamos')
```

#### PENALES: Dados tres enteros entre 1 y 11 que representan los penales cobrados a favor en tres series distintas, si su suma es menor o igual a 21 devuelve esa suma. Si la suma excede 21 *y* hay un 11 entre los tres números, resta 10 a la suma total. Si aún así la suma excede 21, devuelve <code>'EXCESO'</code>.

    penales(5,6,7) --> 18
    penales(9,9,9) --> 'EXCESO'
    penales(9,9,11) --> 19

```python
def penales(a, b, c):
    suma = a + b + c
    if suma <= 21:
        return suma
    if 11 in (a, b, c):
        suma -= 10
    if suma <= 21:
        return suma
    return 'EXCESO'
```

```python
# Revisión
penales(5, 6, 7)
```

```python
# Revisión
penales(9, 9, 9)
```

```python
# Revisión
penales(9, 9, 11)
```
